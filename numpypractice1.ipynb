```python
#Q91. Remove outliers using IQR
arr = np.array([10, 12, 15, 18, 20, 100])
q1=np.percentile(arr,25)
q3=np.percentile(arr,75)
iqr=q3-q1
result=arr[(arr>=q1-1.5*iqr)&
           (arr<=q3+1.5*iqr)]
print(result)
```

```python
#Q100. Small NumPy Data Analysis
arr = np.array([10, 20, 35, 40, 55, 70, 85])
mean = np.mean(arr)
filtered = arr[arr > mean]
sorted_arr = np.sort(arr)

normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print("Mean:", mean)
print("Above Mean:", filtered)
print("Sorted:", sorted_arr)
print("Normalized:", normalized)
```
