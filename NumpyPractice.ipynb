# Numpy

```python
#1.Create a NumPy array [10, 20, 30, 40, 50] and print it.
import numpy as np
arr=np.array([10,20,30,40,50])
print(arr)
```

```python
#2.Create an array of 5 zeros using NumPy.
arr=np.zeros(5)
print(arr)
```

```python
#3.Create an array of 5 ones
arr=np.ones(5)
print(arr)
```

```python
#4.Create numbers from 1 to 10
arr=np.arange(1,11)
print(arr)
```

```python
#5.Create 5 equally spaced numbers from 0 to 10
arr=np.linspace(0,10,5)
print(arr)
```

```python
#6.Find the shape
arr=np.array([[10,20,30],
             [40,50,60]])
print(arr.shape)
```

```python
#7. Find the number of dimensions
arr=np.array([[10,20],[30,40]])
print(arr.ndim)
```

```python
#8.Find the total number of elements
arr=np.array([10,20,30,40,50])
print(arr.size)
```

```python
#9. Find the data type
arr=np.array([1,2,3,4,5])
print(arr.dtype)
```

```python
#10. Access the 3rd element
arr=np.array([10,20,30,40,50])
print(arr[2])
```

```python
#11. Print the first 3 elements
arr=np.array([10,20,30,40,50])
print(arr[:3])
```

```python
#12. Print the last 3 elements
arr=np.array([10,20,30,40,50])
print(arr[-3:])
```

```python
#13. Reverse a NumPy array
arr=np.array([10,20,30,40,50])
print(arr[::-1])
```

```python
#14.Print every second element
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[::2])
```

```python
#15. Change the 2nd element to 100
arr = np.array([10, 20, 30, 40, 50, 60])
arr[1]=100
print(arr)
```

```python
#16. Extract [20, 30, 40]
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[1:4])
```

```python
#17. Print the first row of a 2D array
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
print(arr[0])
```

```python
#18. Print the second column
arr=np.array([[10,20,30],
              [40,50,60]])
print(arr[:,1])
```

```python
#19. Access a specific element
arr=np.array([[10,20,30],
              [40,50,60]])
print(arr[1,2])
```

```python
#20. Replace one element in a 2D array
arr = np.array([[10, 20, 30],
                [40, 50, 60]])
arr[1,2]=100
print(arr)
```

```python
#21. Add 10 to every element
arr=np.array([10,20,30,40])
print(arr+10)
```

```python
#22. Multiply every element by 2
arr=np.array([10,20,30,40])
print(arr*2)
```

```python
#23. Find the sum of all elements
arr=np.array([10,20,30,40])
print(np.sum(arr))
```

```python
#24. Find maximum and minimum
arr=np.array([10,20,30,40])
print(np.max(arr))
print(np.min(arr))
```

```python
#25. Find the mean
arr=np.array([10,20,30,40])
print(np.mean(arr))
```

```python
#26. Find the median
arr=np.array([10,20,30,40,50])
print(np.median(arr))
```

```python
#27. Find standard deviation
arr=np.array([10,20,30,40,])
print(np.std(arr))
```

```python
#28. Find the sum of each row #axis=1 → calculates row-wise.
arr=np.array([[10,20,30],
             [40,50,60]])
print(np.sum(arr,axis=1))
```

```python
#29. Find the sum of each column #axis=0 → calculates column-wise.
arr=np.array([[10,20,30],
             [40,50,60]])
print(np.sum(arr,axis=0))
```

```python
#30. Find the square of every element
arr=np.array([1,2,3,4,5])
print(arr**2)
```

```python
#31. Find numbers greater than 25
arr = np.array([10, 20, 30, 40, 50])
array=arr[arr>25]
print(array)
```

```python
#32. Find all even numbers
arr = np.array([10,15, 20,25, 30,35,])
array=arr[arr%2==0]
print(array)
```

```python
#33. Find all odd numbers
arr = np.array([10,15, 20,25, 30,35,])
array=arr[arr%2!=0]
print(array)
```

```python
#34. Replace values greater than 50 with 0
arr=np.array([30,40,50,60,70,80,90])
arr[arr>50]=0
print(arr)
```

```python
#35. Count values greater than 30
arr=np.array([20,30,40,50,60])
count=np.sum(arr>30)
print(count)
```

```python
#36. Use np.where()
#If value is greater than 50 → "High"Otherwise → "Low"
arr=np.array([30,40,50,60,70])
result=np.where(arr>50,"High","Low")
print(result)
```

```python
#37. Find unique values
arr=np.array([10,30,10,20,20,30,40])
result=np.unique(arr)
print(result)
```

```python
#38. Sort an array
arr=np.array([30,20,50,70,10])
result=np.sort(arr)
print(result)
```

```python
#39. Find the index of maximum value
arr=np.array([30,20,50,70,10])
result=np.argmax(arr)
print(result)
```

```python
#40. Find the index of minimum value
arr=np.array([30,20,50,70,10])
result=np.argmin(arr)
print(result)
```

```python
#41. Reshape an array into 3 × 4
arr=np.arange(1,13)
result=arr.reshape(3,4)
print(result)
```

```python
#42. Convert 2D array into 1D using flatten()
arr=np.array([[1,2,3],
              [4,5,6]])
result=arr.flatten()
print(result)
```

```python
#43. Use ravel()         
arr = np.array([[1,2,3],
              [4,5,6]])        
result = arr.ravel()
print(result)
```

```python
#44. Use broadcasting
arr = np.array([[1,2,3],
              [4,5,6]])        
result=arr+np.array([10,20,30])
print(result)
```

```python
#45. Find unique values and their counts
arr = np.array([10, 20, 10, 30, 20, 10])
values,counts=np.unique(arr,return_counts=True)
print(values)
print(counts)
```

```python
#46. Find common elements between two arrays
arr1=np.array([10,20,30,40])
arr2=np.array([30,40,50,60])
result=np.intersect1d(arr1,arr2)
print(result)
```

```python
#47. Combine two arrays
arr1=np.array([10,20,30,40])
arr2=np.array([30,40,50,60])
result=np.concatenate((arr1,arr2))
print(result)
```

```python
#48. Split an array into 3 parts
arr = np.array([10, 20, 10, 30, 20, 10])
result=np.split(arr,3)
print(result)
```

```python
#49. Generate 10 random integers
arr=np.random.randint(1,101,10)
print(arr)
```

```python
#50. Find the average of values greater than 50
arr = np.array([30,40,60,90,80,70,50])
result=np.mean(arr[arr>50])
print(result)
```
