```python
#Q82. Find the second largest digit
def second_largest_digit(s):
    digits=[]
    for item in s:
        if item.isdigit() and int(item) not in digits:
            digits.append(int(item))
    digits.sort(reverse=True)
    return digits[1]

s = "a523b8"
second_largest_digit(s)    
```
