```python
#Q19. Combine Two Dictionaries by Adding Values
#Write a Python function to combine two dictionaries by adding values for common keys.
def combine_two_dictionary(d1,d2):
    dic={}
    for key in d1:
        dic[key]=d1[key]

    for key in d2:
        if key in dic:
            dic[key]+=d2[key]
        else:
            dic[key]=d2[key]
    return dic 

d1 = {"a":100, "b":200, "c":300}
d2 = {"a":300, "b":200, "d":400}
combine_two_dictionary(d1,d2)
    
        
    
```

```python
#Q46. Group Key-Value Pairs into Dictionary of Lists
#Write a Python program to create a dictionary grouping a sequence of key-value pairs into a
#dictionary of lists.
def grouping_sequence_keyvalue_dic_lst(lst):
    dic={}
    for key,value in lst:
        if key in dic:
            dic[key].append(value)
        else:
            dic[key]=[value]
    return dic 
lst = [("yellow", 1),("blue", 2),("yellow", 3),("blue", 4),("red", 1)]
grouping_sequence_keyvalue_dic_lst(lst)   
```

```python
#Q47. Split Dictionary of Lists into List of Dictionaries
#Write a Python program to split a dictionary of lists into a list of dictionaries.
def split_dictionary(dic):
    result=[]
    length=len(next(iter(dic.values())))
    for i in range(length):
        new_dic={}
        for key in dic:
            new_dic[key]=dic[key][i]
        result.append(new_dic)
    return result  
dic = {"Science": [88, 89, 62, 95],
       "Language": [77, 78, 84, 80]}
split_dictionary(dic)    
```

```python
#Q48. Remove Specified Dictionary from a List
#Write a Python program to remove a specified dictionary from a given list.
def remove_dictionary(lst,remove_id):
    new_lst=[]
    for dic in lst:
        if dic["id"]!=remove_id:
            new_lst.append(dic)
    return new_lst
lst = [{"id": "#FF0000", "color": "Red"},
    {"id": "#800000", "color": "Maroon"},
    {"id": "#FFFF00", "color": "Yellow"},
    {"id": "#808000", "color": "Olive"}]
remove_dictionary(lst,"#800000")    
```

```python
#Q49. Convert String Values into Numeric Types
#Write a Python program to convert string values of dictionaries into integer values.
def convert_string_value_into_numeric_value(lst):
    for dic in lst:
        for key in dic:
            dic[key]=int(dic[key])

    return lst  
lst = [
    {"x": "10", "y": "20", "z": "30"},
    {"p": "40", "q": "50", "r": "60"}
]
convert_string_value_into_numeric_value(lst)
    
```

```python
#Q50. Clear List Values in a 
#Write a Python program to clear the list values in a dictionary.
def clear_list_values_dic(dic):
    for key in dic:
        dic[key]=[]
    return dic
dic = {
    "C1": [10, 20, 30],
    "C2": [20, 30, 40],
    "C3": [12, 34]
}
clear_list_values_dic(dic)
```
