```python
#Q1. Find all even numbers from a tuple
def find_even_number(t):
    result=()
    for item in t:
        if item%2==0:
            result=result+(item,)
    return result
    
t = (10, 15, 20, 25, 30)
find_even_number(t)
```

# Level 3

```python
#Q21. Remove duplicates
def remove_duplicate(t):
    result=()
    for item in t:
        if item not in result:
            result=result+(item,)
    return result         
        
t = (10, 20, 10, 30, 20, 40)
remove_duplicate(t)
```

```python
#Q22. Find repeated elements
def repeated_element(t):
    result=()
    for item in t:
        if t.count(item)>1 and item not in result:
            result=result+(item,)
    return result
    
t = (10, 20, 10, 30, 20, 40)
repeated_element(t)        
```

```python
#Q23. Find elements occurring only once/single element
def single_element(t):
    result=()
    for item in t:
        if t.count(item)==1:
            result=result+(item,)
    return result
    
t = (10, 20, 10, 30, 20, 40)
single_element(t)           
```

```python
#Q24. Find second largest
def second_largest_element(t):
    values=()
    for item in t:
        if item not in values:
            values=values+(item,)
    value=sorted(values,reverse=True)
    return value[1]

t = (10, 50, 20, 50, 30)
second_largest_element(t)    
```

```python
#Q28. Difference between largest and smallest
def difference(t):
    largest=t[0]
    smallest=t[0]
    for item in t:
        if item > largest:
            largest=item
        if item < smallest:
            smallest=item
    return largest,smallest
    
t = (20, 50, 10, 80, 30)
difference(t)        
```

```python
#Q29. Find average without sum()
def average(t):
    total=0
    for item in t:
        total+=item
    return total/len(t)
t = (10, 20, 30, 40)
average(t)   
```

```python
#Q30. Find the most frequent element/most repeated
def most_frequent_element(t):
    element=t[0]
    highest=0
    for item in t:
        count=t.count(item)
        if count>highest:
            highest=count
            element=item
    return element

t = (10, 20, 10, 30, 10, 20)
most_frequent_element(t)    
```

```python
#Q34. Find unique elements from two tuples
def unique_element(t1,t2):
    combined=t1+t2
    result=()
    for item in combined:
        if combined.count(item)==1:
            result=result+(item,)
    return result

t1 = (10, 20, 30)
t2 = (30, 40, 50)
unique_element(t1,t2)
```

```python
#Q37. Find pairs whose sum equals target
def pairs_with_sum(t,target):
    result=()
    for i in range(len(t)):
        for j in range(i+1,len(t)):
            if t[i]+t[j]==target:
                result=result+((t[i],t[j]),)
    return result

t = (10, 20, 30, 40, 50)
pairs_with_sum(t,60)
```

```python
#Q39. Pair having the smallest sum
def smallest_pair_sum(t):
    smallest=None
    pair=()
    for i in range(len(t)):
        for j in range(i+1,len(t)):
            total=t[i]+t[j]
            if smallest is None or total<smallest:
                 smallest=total
                 pair=(t[i],t[j])
    return pair
    
t = (10, 20, 30, 40)
smallest_pair_sum(t)
```

```python
#Q40. Check whether tuple is sorted
def wether_tuple_sorted(t):
    for i in range(len(t)-1):
        if t[i]>t[i+1]:
            return False
    return True

t = (10, 20, 30, 40, 50)
wether_tuple_sorted(t)
```
