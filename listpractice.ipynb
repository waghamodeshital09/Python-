```python
#21. Separate even and odd numbers into two different lists.
def separate_even_odd(lst):
    even = []
    odd = []
    for i in lst:
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)
    return even, odd

lst = [10, 15, 22, 33, 40]

separate_even_odd(lst)
```

```python
def remove_duplicate(lst):
    unique=[]
    for i in lst:
        if i not in unique:
            unique.append(i)
    return unique


lst=[10,20,20,30,40,30]
remove_duplicate(lst)
```

```python
def prime_number(lst):
    prime=[]
    for n in lst:
        count=0
        for i in range(1,n+1):
            if n%i==0:
                count+=1
        if count==2:
            prime.append(n)
    return prime



lst=[2,4,5,8,11,15]
prime_number(lst)
```

```python
def count_prime(lst):
    count=0
    for n in lst:
        factors=0
        for i in range(1,n+1):
            if n%i==0:
                factors+=1
        if factors==2:
            count+=1
    return count

lst = [2, 4, 5, 8, 11, 15]
count_prime(lst)
```

```python
#26. Find the maximum difference between two consecutive elements
def max_difference(lst):
    maximum=0
    for i in range(len(lst)-1):
        diff=abs(lst[i]-lst[i+1])
        if diff>maximum:
            maximum=diff
    return maximum

lst=[10,15,30,40,60]
max_difference(lst)
```

```python
#27. Rotate the list to the left by one position.
def rotate_left(lst):
    first = lst[0]

    for i in range(len(lst) - 1):
        lst[i] = lst[i + 1]

    lst[-1] = first

    return lst


lst = [10, 20, 30, 40, 50]

rotate_left(lst)
```

```python
#28. Rotate the list to the right by one position.
def rotate_right(lst):
    last=lst[-1]
    for i in range(len(lst)-1,0,-1):
        lst[i]=lst[i-1]
    lst[0]=last
    return lst

lst = [10, 20, 30, 40, 50]

rotate_right(lst)    
        
                  
```

```python
#29. Merge two lists into one.
def merge_lst(lst1,lst2):
    merge=[]
    for i in lst1:
        merge.append(i)
    for i in lst2:
        merge.append(i)
    return merge


lst1 = [10, 20, 30]
lst2 = [40, 50, 60]

merge_lst(lst1, lst2)        
```

```python
#36. Count the number of vowels in a list of characters.
def count_vowels(lst):
    count = 0
    for i in lst:
        if i in "aeiouAEIOU":
            count += 1
    return count


lst = ['a', 'b', 'e', 'x', 'i', 'o', 'u']

count_vowels(lst)
```

```python
#38. Find the shortest string in a list.
def shortest_string(lst):
    shortest = lst[0]
    for i in lst:
        if len(i) < len(shortest):
            shortest = i
    return shortest

lst = ["cat", "elephant", "dog", "apple"]
shortest_string(lst)
```
